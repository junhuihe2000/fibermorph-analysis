```python
import numpy as np
from scipy.interpolate import make_splrep
```

```python
theta = np.linspace(0, np.pi, 500)
r = 2 # radius
coords = r * np.array([np.cos(theta), np.sin(theta)]).T
print(f"curvature: {1/r}")
```

```python
# fit a spline to the coordinates
spl_x = make_splrep(theta, coords[:, 0], s=len(theta))
spl_y = make_splrep(theta, coords[:, 1], s=len(theta))

# calculate the first and second derivatives of the spline
dx = spl_x.derivative(1)(theta)
dy = spl_y.derivative(1)(theta)
ddx = spl_x.derivative(2)(theta)
ddy = spl_y.derivative(2)(theta)
# calculate the curvature
curvature = np.abs(dx * ddy - dy * ddx) / (dx ** 2 + dy ** 2) ** (3 / 2)
```

```python
print(f"mean curvature: {np.mean(curvature): .2f}, std curvature: {np.std(curvature): .2f}")
```
