## 2. Curvature calculation
For each single hair fragment, we first fit a principal curve to its coordinates for denoising. Then the curvature is calculated based on this estimated curve. 

### 2.1. Principal curves
Principal curves are one-dimensional smooth curves that best approximate noisy points. In this study, we fit a principal curve using ISOMAP and smoothing splines. The approach includes the following major steps:
1. Suppose the 2D coordinates of one single hair fragment are given as $\{x_i,y_i\}_{i=1}^m$.
2. The ISOMAP algorithm is used for nolinear dimension reduction to obtain the 1D representation $\{t_i\}_{i=1}^m$.
3. We employ smoothing splines to estimate the parametric equation $x(t)$ and $y(t)$.
4. The principal curve is defined as $(x(t), y(t))$.

### 2.2. Curvature formula
Rather than using the circle-fitting algorithm, we direct employ the formula of curvature to calculate the curvature. Given the parametric equation $(x(t), y(t))$, the curvature formula is given as,
$$\kappa(t) = \frac{|x'(t)y''(t) - y'(t)x''(t)|}{|x'(t)^2 + y'(t)^2|^{3/2}}.$$
The mean curvature is given by,
$$\bar{\kappa} = \frac{1}{m} \sum_{i=1}^m \kappa(t_i).$$
When the hair fragment is not approximately a circle, this formula will offer an more accurate estimation for curvature compared to circle-fitting.
